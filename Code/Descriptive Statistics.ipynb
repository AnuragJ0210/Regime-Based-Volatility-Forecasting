```
import pandas as pd 
import numpy as np
from scipy.stats import describe 

df = pd.read_csv("X.csv")

df["Date"] = pd.to_datetime(df["Date"])
mask = (df["Date"] >= "2020-01-01") & (df["Date"] <= "2021-01-01")
df_filtered = df[mask]

y = df_filtered["RV_day"].dropna()
result = describe(y)

print(result)
```

```
import pandas as pd
from sklearn.preprocessing import StandardScaler

y_raw = pd.read_csv("y.csv") 

dates = y_raw["Date"]
y_values = y_raw.drop(columns=["Date"])

scaler = StandardScaler()
y_scaled_values = scaler.fit_transform(y_values)


y_scaled = pd.DataFrame(y_scaled_values, columns=y_values.columns, index=y.index)
y_scaled.insert(0, "Date", dates)

y_scaled.to_csv("y_scaled.csv", index=False)

print("Means:", scaler.mean_)
print("Stds:", scaler.scale_)
```

```
import joblib

# Save scaler
joblib.dump(scaler, 'y_scaler.pkl')

# Load scaler later
scaler_loaded = joblib.load('y_scaler.pkl')
```
