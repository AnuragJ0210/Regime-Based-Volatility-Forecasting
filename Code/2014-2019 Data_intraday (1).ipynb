```
import pandas as pd
import numpy as np

# Read all CSV files from 4 to 22 and concatenate them vertically
file_paths = [f"/work/spx_intraday_2014.xlsx - {i}.csv" for i in range(4, 23)]

# Read and concatenate
all_dataframes = [pd.read_csv(file) for file in file_paths]
SPX_2014_df = pd.concat(all_dataframes, ignore_index=True)

SPX_2014_df
```

```
SPX_2014_df = SPX_2014_df[~SPX_2014_df['Date'].str.contains('16:05|16:10|16:15')]
SPX_2014_df
```

```
# Convert 'Date' column to datetime for proper sorting
SPX_2014_df['Date'] = pd.to_datetime(SPX_2014_df['Date'], format='%m/%d/%y %H:%M')
# Sort the dataframe in chronological order
SPX_2014_df = SPX_2014_df.sort_values(by='Date')
SPX_2014_df
```

```
SPX_2014_df = SPX_2014_df[['Date', 'Close']]
SPX_2014_df
```

```
import numpy as np
SPX_2014_df['ln(pt/pt-1)'] = np.log(SPX_2014_df['Close'] / SPX_2014_df['Close'].shift(1))
SPX_2014_df = SPX_2014_df.dropna()
SPX_2014_df = SPX_2014_df[~(SPX_2014_df['Date'].dt.time == pd.to_datetime("09:30").time())]
SPX_2014_df
```

```
SPX_2014_df.set_index('Date', inplace=True)
```

```
intraday_2014 = pd.DataFrame(SPX_2014_df['ln(pt/pt-1)'])
```

```
intraday_2014.sort_index(inplace=True)
```

```
intraday_2014.to_csv("intraday_2014.csv", index=True)
```

```
SPX_2014_df["Date"] = SPX_2014_df.index
```

```
SPX_2014_df
```

```
# Resample SPX_close_df to daily frequency, summing up measurements, dividing by the number of measurements, and multiplying by 78
SPX_daily_2014_df = SPX_2014_df.resample('D', on='Date').apply(
    lambda x: np.sqrt((x**2).sum() * 78 / len(x))
)

# Resample VIX_close_df to daily frequency, summing up measurements, dividing by the number of measurements, and multiplying by 78

SPX_daily_2014_df = pd.DataFrame(SPX_daily_2014_df)
SPX_daily_2014_df = SPX_daily_2014_df.dropna(subset=['ln(pt/pt-1)']) 
SPX_daily_2014_df = SPX_daily_2014_df[['ln(pt/pt-1)']].rename(columns={'ln(pt/pt-1)': 'RV'})
SPX_daily_2014_df
```

```
SPX_daily_2014_df.sort_index(inplace=True)
SPX_daily_2014_df.index.name = "Date"
SPX_daily_2014_df.to_csv("SPX_daily_2014.csv", index=True)
SPX_daily_2014_df
```
